```python
df["Physical_Activity_Hours"] = df["Physical_Activity_Hours"].clip(lower = 0)
```

```python
num_cols = df.select_dtypes(include = "number")
num_cols.skew()
```

```python
top_countries = df['Country'].value_counts().index[:10].tolist()
```

```python
def group_countries(country):
  if country in top_countries:
    return country
  else:
    return 'Other'
```

```python
df['Grouped_country'] = df['Country'].apply(group_countries)
```

```python
df['Grouped_country'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
skewed_col = ['Study_Hours'] #as it is skewed column we need to perform log transformation

other_numeric_cols = [   # for all the remaining muneric columns we need to perform standard scaling as it narrows down in a range for better model performance
    "Age",
    "Avg_Daily_Usage_Hours",
    "Daily_Unlocks",
    "Physical_Activity_Hours",
    "Sleep_Hours_Per_Night"
]

ordinal_col = ['Stress_Level'] # it is a categorical column as in this column the order and the priority matter we perfor ordinal(label) encoding to maintain the order

nominal_col = [ # and these are also the categorical cols but the order and priority doesnt matter so we perform the one hot encoding for this column
    "Gender",
    "Academic_Level",
    "Most_Used_Platform",
    "Purpose_Of_Use",
    "Grouped_country"
]

feature_col = skewed_col + other_numeric_cols + ordinal_col + nominal_col

X = df[feature_col]
y = df['Mental_Health_Score']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    FunctionTransformer,
    StandardScaler,
    OrdinalEncoder,
    OneHotEncoder
)
from sklearn.compose import ColumnTransformer


# 1. Skewed Features
skew_pipeline = Pipeline(steps=[
    ('log_transform', FunctionTransformer(np.log1p)),
    ('scale', StandardScaler())
])


# 2. Numeric Features
plain_numeric_pipeline = Pipeline(steps=[
    ('scale', StandardScaler())
])


# 3. Ordinal Features
ordinal_pipeline = Pipeline(steps=[
    ('encode', OrdinalEncoder(
        categories=[['Low', 'Medium', 'High', 'Very High']]
    ))
])


# 4. Nominal Features
nominal_pipeline = Pipeline(steps=[
    ('encode', OneHotEncoder(handle_unknown="ignore"))
])


# Combine all preprocessing pipelines
preprocessor = ColumnTransformer(transformers=[
    ("Skewed_Pipeline", skew_pipeline, skewed_col),
    ("Plain_Numeric", plain_numeric_pipeline, other_numeric_cols),
    ("Ordinal", ordinal_pipeline, ordinal_col),
    ("Nominal", nominal_pipeline, nominal_col)
])
```

```python
from sklearn.linear_model import LinearRegression

lr_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])

lr_pipeline.fit(X_train, y_train)

lr_preds = lr_pipeline.predict(X_test)
```

```python
lr_preds
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

mae = mean_absolute_error(y_test, lr_preds)
mse = mean_squared_error(y_test, lr_preds)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, lr_preds)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
lr_preds_train = lr_pipeline.predict(X_train)
lr_r2_train = r2_score(y_train, lr_preds_train)
print(lr_r2_train)
```

```python
test_preds = lr_pipeline.predict(X_test)
test_r2 = r2_score(y_test, test_preds)
print(test_r2)
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', RandomForestRegressor(random_state=42))
])

rf_pipeline.fit(X_train, y_train)
rf_preds = rf_pipeline.predict(X_test)
```

```python
rf_r2_test = r2_score(y_test, rf_preds)
print(rf_r2_test)
```

```python
rf_pred_train = rf_pipeline.predict(X_train)
rf_r2_train  = r2_score(y_train, rf_pred_train)
print(rf_r2_train)
```

```python
mae = mean_absolute_error(y_test, rf_preds)
mse = mean_squared_error(y_test, rf_preds)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, rf_preds)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    'regressor__n_estimators': [100, 200, 300],
    'regressor__max_depth': [5, 10, 15],
    'regressor__min_samples_split': [2, 5, 10],
    'regressor__min_samples_leaf': [1, 2, 4]
}

random_search = RandomizedSearchCV(
    estimator=rf_pipeline, #as pipeline is created if not randomforestregressor
    param_distributions=param_grid,
    n_iter=15,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1
)
```

```python
random_search.fit(X_train, y_train)
```

```python
random_search.best_params_
```

```python
rf_best_pipeline = random_search.best_estimator_
rf_best_pipeline
```

```python
rf_best_preds = rf_best_pipeline.predict(X_test)
rf_best_r2 = r2_score(y_test, rf_best_preds)
print(rf_best_r2)
```
